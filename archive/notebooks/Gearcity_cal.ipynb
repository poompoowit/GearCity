```python
results = []
for i in range(len(df)):
    # Calculate scores for all possible demographic combinations
    best_score = float('-inf')
    best_demographics = None
    for gender in ['Male', 'Female', 'Neutral']:
        for age in ['Less Than 25','25-35','35-55','Greater Than 55']:
            # Calculate score for this demographic combination
            score = 0
            for col in df.columns:
              if col != 'Vehicle Type':
                score += df.iloc[i][col]*(1+demographics_gender[gender].get(col,0)+demographics_age[age].get(col,0))

            # Update best score and demographics if necessary
            if score > best_score:
                best_score = score
                best_demographics = {'gender': gender, 'age': age}

    # Store results for this vehicle type
    results.append({'Vehicle Type': df.iloc[i]['Vehicle Type'], 'Best Demographics Gender': best_demographics['gender'], 'Best Demographics Age': best_demographics['age'], 'Best Score': best_score})

# Create a DataFrame from the results
result_df = pd.DataFrame(results)
```
